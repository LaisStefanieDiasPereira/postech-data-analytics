# Create PySpark DataFrame From an Existing RDD

RDD (Resilient Distributed Dataset) é a estrutura de dados fundamental do Apache Spark.
Ele representa uma coleção de dados distribuída em múltiplos nós do cluster, permitindo processamento paralelo e tolerância a falhas.

🔑 Principais características:

Resiliente → tolerante a falhas, pois se uma partição for perdida, o Spark pode reconstruí-la usando o histórico das operações (lineage).

Distribuído → os dados são particionados em diferentes máquinas, aproveitando o poder do cluster.

Dataset (coleção de objetos) → pode conter qualquer tipo de objeto do Scala, Java ou Python.

Imutável → uma vez criado, não pode ser alterado. Em vez disso, criamos novos RDDs a partir de transformações.

Avaliação preguiçosa (lazy evaluation) → operações de transformação só são executadas quando uma ação (action) é chamada.

⚙️ Tipos de operações em RDDs:

Transformações → criam novos RDDs a partir de outros (ex: map, filter, flatMap, reduceByKey).
Elas são lazy, ou seja, não rodam imediatamente.

Ações → retornam resultados ou salvam dados (ex: collect, count, saveAsTextFile).
Elas disparam a execução das transformações.

```python
rdd = sc.parallelize([('C',85,76,87,91), ('B',85,76,87,91), ("A", 85,78,96,92), ("A", 92,76,89,96)], 4) # Parallelize é praticamente uma bibliotreca que usamos para inicializar o RDD
```

```python
print(type(rdd)) # Definimos o tipo de objeto que acabamos de criar
```

```python
sub = ['Division','English','Mathematics','Physics','Chemistry'] # renomeia as colunas criadas no parallelize
marks_df = spark.createDataFrame(rdd, schema=sub) #informa qual o skema de dadis da nossa tabela
```

```python
print(type(marks_df)) # para visualizar o tipo de arquivo, no caso criamos um data frame em cima de um rdd existente
```

```python
marks_df.printSchema() #nulabble significa que valor dessa coluna pode ser nulo
```

```python
marks_df.show() # para visualizar os dados
```
