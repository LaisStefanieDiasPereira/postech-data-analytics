# Creating New Column in PySpark DataFrame

## Installing Libraries, Importing the Libraries and Creating Spark Session

```python
!pip install pyspark
import pyspark
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("pysparkdf").getOrCreate()
```

## Importing Data

```python
df = spark.read.csv('cereal.csv', sep = ',', inferSchema = True, header = True)
```

## Reading the Schema

```python
df.printSchema() # para visualizar o que contemos dentro do conteudo
```

# select()

```python
df.select('name', 'mfr', 'rating').show(10) # Select para selecionar colunas / informações
```

# withColumn()

```python
df.withColumn("Calories",df['calories'].cast("Integer")).printSchema() # fazer isso para substituir um campo e renomer ele exemplo de calories para Calories
```

# groupBy()

```python
df.groupBy("name", "calories").count().show() # essa função agrupa os campos sinalizados contando e apresenta as informações de duplicidade, porém precisa passar a função conforme ela é escrita
```

```python
df.groupBy("calories").count().show() # essa função agrupa os campos sinalizados contando e apresenta as informações de duplicidade, porém precisa passar a função conforme ela é escrita
```

# orderBy()

```python
df.orderBy("protein").show() # ordena a pela coluna sinalizada
```

```python
df.orderBy("calories").show() # ordena a pela coluna sinalizada
```

Diferença do cas when do SQL e caso When do pyspark???

# when()

```python
from pyspark.sql.functions import when
```

```python
df.select("name", when(df.vitamins >= "25", "rich in vitamins")).show() # coloca o nome das colunas que deseja visualizr e a criar como se fosse um substituir, no caso quando vitamins for maior ou igual a 25 é rico em vitaminas
```

# filter()

```python
df.filter(df.calories == "120").show() # Filtrar igual a
```

```python
df.filter(df.calories >= "120").show() # Filtro de maior ou igual a
```

```python
df.filter(df.calories != "120").show() # filtro diferente de
```

# isNull()/isNotNull()

```python
#isNotNull()
from pyspark.sql.functions import * # * é para instalar todas as funções bibliotecas para a sessão.
#filter data by null values
df.filter(df.name.isNotNull()).show() # trouxe todas as linhas que não são nulas para este data frame
```

# isNull()

```python
df.filter(df.name.isNull()).show() # trouxe todas as linhas que são nulas para este data frame
```
